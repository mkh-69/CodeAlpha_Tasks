```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
```

```python
df = pd.read_excel("default of credit card clients.xls", header=1)

df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
df
```

```python
print(df["default payment next month"].value_counts())
```

```python
df["TOTAL_BILL_AMT"] = (
    df["BILL_AMT1"] +
    df["BILL_AMT2"] +
    df["BILL_AMT3"] +
    df["BILL_AMT4"] +
    df["BILL_AMT5"] +
    df["BILL_AMT6"]
)
```

```python
df["TOTAL_PAY_AMT"] = (
    df["PAY_AMT1"] +
    df["PAY_AMT2"] +
    df["PAY_AMT3"] +
    df["PAY_AMT4"] +
    df["PAY_AMT5"] +
    df["PAY_AMT6"]
)
```

```python
df[["TOTAL_BILL_AMT", "TOTAL_PAY_AMT"]].head()
```

```python
X = df.drop("default payment next month", axis=1)
y = df["default payment next month"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("Training data:", X_train_scaled.shape)
print("Testing data:", X_test_scaled.shape)
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
y_pred
```

```python
print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```

```python
y_probability = model.predict_proba(X_test_scaled)[:, 1]

roc_auc = roc_auc_score(y_test, y_probability)

print("ROC-AUC:", roc_auc)
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Default", "Default"]
)

disp.plot()
plt.title("Credit Scoring Confusion Matrix")
plt.show()
```

```python
RocCurveDisplay.from_predictions(y_test, y_probability)

plt.title("ROC Curve - Credit Scoring Model")
plt.show()
```
